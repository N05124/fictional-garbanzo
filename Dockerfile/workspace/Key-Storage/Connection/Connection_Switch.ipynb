```python
# Dispatch manager
def status_response(process):
    match process['node']:
        case "origin":
            process["status"] = case_origin(process)

        case "pivot":
            process["status"] = case_pivot(process)

        case "storage":
            process["status"] = case_storage(process)

        case _:
            raise ValueError(f"Unknown status: {process}")
    return process
    
status_response("origin")
```

```python
def case_storage(status):

    match status:

        # allocation is sufficient
        case "allocate":
            return None

        # session is valid
        case "init_session":
            return None

        # blockchain availability
        case {"available": is_available}:

            match is_available:

                case "init_block":
                    return None

                case "symetric":
                    return None

                case "o_ping":
                    return None

        # chain integrity
        case {"integrity": integrity}:

            match integrity:

                case "init_block":
                    return None

                case "pre-block-chain":
                    return None

                case "post-block-chain":
                    return None

                case "credential":
                    return None

        # storage test successful
        case "test-storage":
            return None

        # storage complete
        case "complete-storage":
            return None

        # session hash complete
        case "hash-session":
            return None

        # chain block complete
        case "chain-block":
            return None

        # close session
        case "close-session":
            return None

        # wait for interrupt
        case "wait4interupt":
            return None

        # final ping
        case "final-ping":
            return None

        # server unavailable
        case "island":
            return None

        # sleeping
        case "sleeping":
            return None

        case _:
            raise ValueError(f"Unknown status: {status}")


case_storage("allocate")
```
