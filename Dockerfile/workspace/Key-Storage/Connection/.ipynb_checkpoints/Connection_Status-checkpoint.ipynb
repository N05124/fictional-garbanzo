```python
# this will need to collect informatoin from the Host and relay to the client. the current status 
# as produced by the process object.
def send_status(process, task):
    try:
        response = requests.post(process['status'], json=task)
        response.raise_for_status()  # raises on 4xx/5xx

        if response.json().get('status') == 'successful':
            return True
        else:
            invalid_attempt('send_status()')
            return False

    except requests.exceptions.RequestException as e:
        invalid_attempt('send_status()')
        print(f"Request failed: {e}")
        return -1
```
