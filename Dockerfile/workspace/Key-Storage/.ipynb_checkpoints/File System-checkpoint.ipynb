```python
import os, sys

# show the actual directory structure
for root, dirs, files in os.walk(".."):
    dirs[:] = [d for d in dirs if d not in ['__pycache__', '.git', '.ipynb_checkpoints', 'node_modules']]
    level = root.replace("..", "").count(os.sep)
    indent = "  " * level
    print(f"{indent}{os.path.basename(root)}/")
    for f in files:
        print(f"{'  ' * (level+1)}{f}")

print("\n--- sys.path ---")
for p in sys.path:
    print(p)
```
